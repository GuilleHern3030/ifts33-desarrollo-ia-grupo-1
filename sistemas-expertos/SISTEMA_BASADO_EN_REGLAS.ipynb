
# Sistema basado en reglas

## Introducción práctica

En este Colab vamos a construir un sistema basado en reglas aplicado al ámbito de la construcción.

Vamos a trabajar con los siguientes conceptos:

- Hechos
- Reglas
- Base de conocimiento
- Motor de inferencia
- Resultado o acción recomendada

La estructura general será:

**HECHOS + REGLAS → MOTOR DE INFERENCIA → CONCLUSIÓN / ACCIÓN**

## 1. El problema

Una empresa constructora quiere contar con un sistema sencillo que ayude a detectar algunas situaciones relacionadas con una obra.

El sistema recibirá información sobre:

- nivel de humedad;
- cantidad/tamaño de fisuras.

A partir de esos datos, utilizará reglas previamente definidas para generar recomendaciones.

### Importante

Este sistema no aprende las reglas automáticamente.

Las reglas fueron definidas previamente a partir del conocimiento del dominio.

```python
# Datos que recibirá nuestro sistema

humedad = 80
fisuras = 2

print("Humedad:", humedad, "%")
print("Fisuras:", fisuras, "mm")
```

¿Qué representan estos valores?
Son hechos/datos conocidos por el sistema.

## 2. Nuestra primera regla

Una regla tiene la estructura:

**SI condición → ENTONCES acción**

Ejemplo:

**SI humedad > 70 → ENTONCES revisar impermeabilización**

La condición determina cuándo se activa la regla.

La acción indica qué debe recomendar el sistema.

```python
if humedad > 70:
    print("Revisar impermeabilización")
```

```python
humedad = 50

if humedad > 70:
    print("Revisar impermeabilización")
```

## 3. Varias reglas

Ahora nuestro sistema tendrá más de una regla.

Regla 1:
SI humedad > 70 → revisar impermeabilización.

Regla 2:
SI fisuras > 5 → revisar estructura.

Regla 3:
SI humedad <= 70 Y fisuras <= 5 → mantenimiento preventivo.

```python
humedad = 80
fisuras = 2

if humedad > 70:
    print("Revisar impermeabilización")

if fisuras > 5:
    print("Reforzar estructura")

if humedad <= 70 and fisuras <= 5:
    print("Todo en condiciones, solo mantenimiento preventivo")
```

## 4. De las reglas a una base de conocimiento

En un sistema basado en reglas no queremos tener todas las reglas mezcladas con el resto del programa.

Podemos organizar las reglas en una **base de conocimiento**.

Cada regla tendrá:

- una condición;
- una acción.

La base de conocimiento contiene el conocimiento que utilizará el sistema.

```python
base_conocimiento = [

    {
        "condicion": lambda humedad, fisuras: humedad > 70,
        "accion": "Revisar impermeabilización"
    },

    {
        "condicion": lambda humedad, fisuras: fisuras > 5,
        "accion": "Reforzar estructura"
    },

    {
        "condicion": lambda humedad, fisuras: humedad <= 70 and fisuras <= 5,
        "accion": "Todo en condiciones, solo mantenimiento preventivo"
    }
]
```

### ¿Qué representa cada elemento?

`condicion` → determina cuándo se activa la regla.

`accion` → indica qué debe hacer o recomendar el sistema.

Cada diccionario representa una regla.

La lista completa representa nuestra base de conocimiento.

```python
for regla in base_conocimiento:
    print(regla)
```

La base de conocimiento no está tomando ninguna decisión todavía. Simplemente contiene las reglas

## 5. Motor de inferencia

Ahora necesitamos un mecanismo que utilice la base de conocimiento.

El motor de inferencia:

1. recibe los hechos;
2. recorre las reglas;
3. evalúa las condiciones;
4. determina qué reglas se cumplen;
5. obtiene las acciones correspondientes.

```python
def motor_inferencia(humedad, fisuras):

    resultados = []

    for regla in base_conocimiento:

        if regla["condicion"](humedad, fisuras):
            resultados.append(regla["accion"])

    return resultados
```

```python
resultado = motor_inferencia(80, 2)

print(resultado)
```

## ¿Qué acaba de pasar?

El sistema recibió:

- humedad = 80
- fisuras = 2

El motor de inferencia recorrió las reglas.

Evaluó sus condiciones.

La regla:

**SI humedad > 70 → revisar impermeabilización**

se cumplió.

Por eso el sistema obtuvo como resultado:

**Revisar impermeabilización**

```python
resultado = motor_inferencia(50, 8)

print(resultado)
```

```python
resultado = motor_inferencia(50, 2)

print(resultado)
```

Hoy construimos un sistema basado en reglas.

Aprendimos que:

**HECHOS**
→ información disponible

**REGLAS**
→ conocimiento representado

**BASE DE CONOCIMIENTO**
→ conjunto de reglas

**MOTOR DE INFERENCIA**
→ evalúa las reglas

**RESULTADO**
→ conclusión o acción

En la próxima clase veremos cómo se realiza el proceso de inferencia cuando tenemos diferentes reglas y cómo se pueden explorar diferentes caminos para resolver un problema.

```python
# Definimos los datos que recibirá el sistema (hechos)
usa_casco = True
usa_zapatos_de_seguridad = True
usa_uniforme_reglamentario = True
```

```python
# Definimos las reglas
if not usa_casco:
    print("Informar al trabajador que debe colocarse el casco")

if not usa_zapatos_de_seguridad:
    print("Informar al trabajador que debe colocarse los zapatos de seguridad")

if not usa_uniforme_reglamentario:
    print("Informar al trabajador que debe colocarse el uniforme reglamentario")

if not usa_casco and not usa_zapatos_de_seguridad and not usa_uniforme_reglamentario:
    print("Suspender al trabajador el resto de la jornada")

if usa_casco and usa_zapatos_de_seguridad and usa_uniforme_reglamentario:
    print("Todo en orden con los elementos de seguridad del trabajador")
```

```python
# Creo la base de conocimiento
base_conocimiento = [

    {
        "condicion": lambda usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario: not usa_casco and not usa_zapatos_de_seguridad and not usa_uniforme_reglamentario,
        "accion": "Suspender al trabajador el resto de la jornada"
    },

    {
        "condicion": lambda usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario: not usa_casco,
        "accion": "Informar al trabajador que debe colocarse el casco"
    },

    {
        "condicion": lambda usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario: not usa_zapatos_de_seguridad,
        "accion": "Informar al trabajador que debe colocarse los zapatos de seguridad"
    },

    {
        "condicion": lambda usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario: not usa_uniforme_reglamentario,
        "accion": "Informar al trabajador que debe colocarse el uniforme reglamentario"
    },

    {
        "condicion": lambda usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario: usa_casco and usa_uniforme_reglamentario and usa_zapatos_de_seguridad,
        "accion": "Todo en orden con los elementos de seguridad del trabajador"
    }
]
```

```python
# Construyo el motor de inferencia
def motor_inferencia(usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario):

    resultados = []

    for regla in base_conocimiento:

        if regla["condicion"](usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario):
            resultados.append(regla["accion"])

    return resultados
```

Casos de prueba

```python
# Caso 1: Sin casco
usa_casco = False
usa_zapatos_de_seguridad = True
usa_uniforme_reglamentario = True

resultado = motor_inferencia(usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario)
print(resultado)
```

📌​ El sistema tomo la decisión de informar al trabajador sobre colocarse el casco porque analizó los hechos y vió que el trabajador no usa casco, lo comparó con las condiciones de las reglas y seleccionó cual aplica, dando el resultado obtenido.

```python
# Caso 2: Sin uniforme reglamentario
usa_casco = True
usa_zapatos_de_seguridad = True
usa_uniforme_reglamentario = False

resultado = motor_inferencia(usa_casco, usa_zapatos_de_seguridad, usa_uniforme_reglamentario)
print(resultado)
```
